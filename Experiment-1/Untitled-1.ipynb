## 1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('employee_information_100.csv')
```

```python
df.info()
```

```python
avg_salary = df.groupby('Department')['Salary'].mean().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10, 6))
avg_salary.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Salary by Department')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.tight_layout()
plt.show()
```

## 1.2: Display the number of employees working in each department.

```python
count_emp  = df.groupby('Department')['Employee_ID'].count().sort_values(ascending=False)
```

```python
print(count_emp)
```

## 1.3:Display the percentage of male and female employees using a pie chart

```python
gender_count = df.groupby('Gender')['Employee_ID'].count().sort_values(ascending=False)
```

```python
print(gender_count)
```

```python
plt.pie(gender_count, labels=gender_count.index, autopct='%1.1f%%', startangle=90, colors=['red' , 'blue'])
plt.title('Gender Distribution of Employees')
plt.axis('equal')
plt.show()
```

## 1.4: Draw a histogram showing salary distribution.

```python
plt.hist(df['Salary'], bins = 10, color='green', edgecolor='black')
plt.title('Salary Distribution of Employees')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.show()
```

## 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Experience_Years"], df["Salary"], color="teal", edgecolor="black")
plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.tight_layout()
plt.show()
```

## 1.6: Display the top 10 employees with the highest salary.

```python
top10 = df.sort_values("Salary", ascending=False).head(10)
top10.head(10)
```

## 1.7: Find the highest salary in every department.

```python
highest_salary = df.groupby('Department')['Salary'].max().sort_values(ascending=False)
```

```python
print(highest_salary)
```

## 1.8: Display employees whose salary is greater than the overall average salary.

```python
avg_salary = df['Salary'].mean()
emp = df[df['Salary'] > avg_salary]
```

```python
print(emp)
```

## 1.9: Calculate the average years of experience in each department.

```python
avg_exp = df.groupby('Department')['Experience_Years'].mean().sort_values(ascending = False)
```

```python
print(avg_exp)
```
